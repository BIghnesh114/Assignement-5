```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sbs

```

```python
plt.style.use("seaborn-v0_8-whitegrid")
```

```python
df=pd.read_csv("student-mat.csv", sep=';')
```

```python
df.head()
```

```python
df.info()
```

```python
dfn = df[['traveltime', 'studytime']]
sbs.lineplot(data=dfn, x='traveltime', y='studytime', dashes=True)
plt.show()
```

#  CORRELATION ANALYSIS


```python
correlation=dfn.corr()
print(correlation)
sbs.heatmap(correlation,cmap="YlGnBu")
plt.show()
```

# covariance analysis

```python
covar=dfn.cov()
print(covar)
sbs.heatmap(covar,cmap="YlGnBu")
plt.show()
```
